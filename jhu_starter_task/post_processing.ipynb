```python
p = "/usr/local/lib/python3.12/dist-packages/monai/transforms/transform.py"
s = open(p).read()
s = s.replace(
    "_seed = _seed % MAX_SEED",
    "_seed = int(_seed) % MAX_SEED"
)
open(p, "w").write(s)
print("patched transform.py")
```

```python
p2 = "/usr/local/lib/python3.12/dist-packages/monai/transforms/compose.py"
s2 = open(p2).read()
s2 = s2.replace(
    'self.R.randint(MAX_SEED, dtype="uint32")',
    'int(self.R.randint(MAX_SEED, dtype="int64"))'
)
open(p2, "w").write(s2)
print("patched compose.py")
```
